### Cell 1: Generate the Scenario
This loads the GPT-2 model and generates a scenario based on a prompt.

```python
from transformers import pipeline

# Load the language model using a direct path to avoid the 401 Unauthorized error
generator = pipeline('text-generation', model='gpt2')

# Generate the scenario
prompt = "Generate a scenario where a company's network faces a ransomware attack and unusual file encryption patterns are detected."
scenario = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, pad_token_id=50256)

print(scenario[0]['generated_text'])
```

### Cell 2: Structure the Incident Data
Convert the scenario into a structured table for analysis.

```python
import pandas as pd

data = {
    'timestamp': ['2023-11-01 12:00:00', '2023-11-01 12:05:00', '2023-11-01 12:10:00'],
    'event_type': ['login_attempt', 'file_encryption', 'network_anomaly'],
    'details': ['suspicious login from IP 192.168.1.5', 'encryption detected on HR folder', 'DDoS attack from multiple IPs']
}

df = pd.DataFrame(data)
print(df)
```

### Cell 3: Detect Anomalies (K-Means)
Use clustering to group normal vs. anomalous events.

```python
from sklearn.cluster import KMeans
import numpy as np

# [1,0] and [0,1] are normal, [3,5] and [4,4] are anomalies
event_data = np.array([
    [1, 0], [0, 1], [1, 1], 
    [3, 5], [4, 4],          
])

kmeans = KMeans(n_clusters=2, random_state=42, n_init='auto')
kmeans.fit(event_data)
labels = kmeans.predict(event_data)

print("Cluster Labels:", labels)
```

### Cell 4: Generate Response Actions
Ask the AI to suggest immediate remediation steps.

```python
response_prompt = "Immediate incident response actions for a ransomware encryption include:"

response_actions = generator(response_prompt, max_length=150, num_return_sequences=1, do_sample=True, pad_token_id=50256, no_repeat_ngram_size=2)

print(response_actions[0]['generated_text'])
```
